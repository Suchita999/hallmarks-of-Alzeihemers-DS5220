# VISUALIZATIONS

```python
# from the meta dataset
# plot 1 - Age Distribution
# plot 2 - Sex Distribution
```

```python
# Histogram for Age distribution
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
sns.histplot(meta['Age'], kde=True, bins=10, color='b')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

```python
# Bar plot for Sex distribution
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 2)
sns.countplot(data=meta, x='Sex')
plt.title('Distribution by sex')
plt.xlabel('Sex')
plt.ylabel('Count')
```

```python
# Tissue
plt.figure(figsize=(14, 10))
sns.countplot(x='Tissue', data=meta)
plt.title('Tissue Distribution')
plt.xlabel('Tissue')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')  # Adjust rotation and alignment of xticks
plt.tight_layout()  # Ensures labels fit within figure area
plt.show()
```

```python
# from Proteins dataset
# plot 1 - Heatmap
```

```python
# Compute the correlation matrix
corr_matrix = proteins.corr()
corr_matrix
```

```python
# Plot heatmap
#plt.figure(figsize=(16, 10))
sns.heatmap(corr_matrix, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Heatmap of Gene Expression")
plt.xlabel("Genes")
plt.ylabel("Samples")
plt.show()
```

```python
# Set the first column (miRNA_IDS) as the index
miRNA.set_index('miRNA_IDS', inplace=True)

# Plot heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(miRNA, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Heatmap of miRNA Expression Levels")
plt.xlabel("Sample")
plt.ylabel("miRNA")
plt.show()
```

```python
# from the mircoarray datatset
# plot 1 - Heatmap
```

```python
microarray.set_index('V1', inplace=True)

# Plot heatmap
#plt.figure(figsize=(14, 10))
sns.heatmap(microarray, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Heatmap of Gene Expression Levels")
plt.xlabel("Sample")
plt.ylabel("Gene")
plt.show()
```

```python
# Plot heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(microarray, cmap="coolwarm")
plt.title("Heatmap of Gene Expression Levels")
plt.xlabel("Sample")
plt.ylabel("Gene")
plt.show()
```

```python
# Plot density plot
plt.figure(figsize=(14, 10))
sns.kdeplot(data=microarray, legend=False)
plt.title('Density Plot of Gene Expression Levels')
plt.xlabel('Expression Level')
plt.ylabel('Density')
plt.show()
```

```python
# Hierarchical clustering of proteins
sns.clustermap(proteins, cmap='viridis', figsize=(12, 10))
plt.title('Clustermap of Protein Expression Levels')
plt.show()
```
